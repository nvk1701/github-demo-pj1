```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

import re
from underthesea import word_tokenize
```

```python

train = pd.read_csv("data/UIT-VSFC_train.csv")
valid = pd.read_csv("data/UIT-VSFC_valid.csv")
test = pd.read_csv("data/UIT-VSFC_test.csv")

print(train.head())
print("Train:", train.shape)
print("Validation:", valid.shape)
print("Test:", test.shape)
```

ĐÁNH GIÁ DỮ LIỆU TỐT HAY CHƯA

```python
train.label.value_counts(normalize=True).sort_index().plot(kind='barh', title='Train label distribution')
```

```python
X_train = train['text']
y_train = train['label']

X_valid = valid['text']
y_valid = valid['label']

X_test = test['text']
y_test = test['label']

print(X_train.head())
print(y_train.head())
```

WORD SEGMENTATION

```python
with open('data/vietnamese-stopwords.txt', 'r', encoding='utf-8') as f:
    stop_words = f.read().splitlines()

def preprocess_text(text):
    # Lowercase the text
    text = text.lower()

    # Xóa các ký tự đặc biệt, dấu câu, chỉ giữ lại chữ cái và số
    text = re.sub(r'[^\w\s]', '', text)

    # Tham số format="text" sẽ nối các từ ghép bằng dấu gạch dưới (VD: "sản phẩm" -> "sản_phẩm")
    text = word_tokenize(text, format="text")

    # Xóa stop words
    words = text.split()
    words = [w for w in words if w not in stop_words]

    return ' '.join(words)

test_text = "Sản phẩm này thật sự rất tuyệt vời, tôi và gia đình rất thích!"
processed_text = preprocess_text(test_text)
print(processed_text)

```

TF-IDF

```python
vectorizer = TfidfVectorizer(tokenizer=preprocess_text, ngram_range=(1, 2), max_features=10000)

vectorizer.fit(X_train)
```

```python
len(vectorizer.vocabulary_)

vectorizer.get_feature_names_out()[200:400]
```

```python
inputs = vectorizer.transform(X_train)
```

```python
inputs.shape

inputs.toarray()[0][:200]
```

```python
test_inputs = vectorizer.transform(X_test)
```

```python
test_inputs.shape
```

```python
X_valid
```

```python
print(X_valid.isna().sum())
print(X_valid[X_valid.isna()])
```

```python
valid_inputs = vectorizer.transform(X_valid)
```

TRAIN LOGICTIS REGRESSION MODEL

```python
model = LogisticRegression()

model.fit(inputs, y_train)
```

```python
train_preds = model.predict(inputs)

train_preds
```

```python
y_train 
```

```python
accuracy_score(y_train, train_preds)
```

```python
valid_preds = model.predict(valid_inputs)

accuracy_score(y_valid, valid_preds)
```

```python
test_preds = model.predict(test_inputs)

accuracy_score(y_test, test_preds)
```
